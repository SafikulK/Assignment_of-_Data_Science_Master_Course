```python
#Q1. Explain with an example each when to use a for loop and a while loop.

#A for loop is used when we know how many times we want to repeat a task.
for i in range(1, 6):
    print(i)
    


```

```python
# A while loop is used when we want to repeat a task while a condition is true.
i = 1

while i <= 5:
    print(i)
    i = i + 1
```

```python
#Q2. Write a python program to print the sum and product of the first 10 natural numbers using for and while loop.
```

```python
sum = 0
product = 1

for i in range(1, 11):
    sum = sum + i
    product = product * i

print("Sum =", sum)
print("Product =", product)
```

```python
#Q4. Create a list of numbers from 1 to 100. Use for loop and while loop to calculate the cube of each number. If the cube is divisible by 4 or 5, append that number to a list and print the list.
```

```python
numbers = []

for i in range(1, 101):
    cube = i * i * i

    if cube % 4 == 0 or cube % 5 == 0:
        numbers.append(i)

print(numbers)
```

```python
numbers = []

i = 1

while i <= 100:
    cube = i * i * i

    if cube % 4 == 0 or cube % 5 == 0:
        numbers.append(i)

    i = i + 1

print(numbers)
```

```python
#Q5. Write a program to filter/count vowels in the given string.
```

```python
string = "I want to become a data scientist"

vowels = "aeiou"
count = 0

for i in string:
    if i.lower() in vowels:
        count = count + 1

print("Number of vowels =", count)
```
